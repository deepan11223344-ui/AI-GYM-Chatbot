# AI-Powered Gym Assistant Chatbot
This notebook demonstrates a conversational AI assistant for gym members using Google Colab and generative AI.

## Features
- Load and analyze gym member exercise tracking data
- Interactive Q&A interface powered by Generative AI
- Real-time insights from member data

```
import pandas as pd
import os
from google.colab import ai

print('Dependencies loaded successfully!')
```

```
# Load the gym dataset
csv_path = '/content/gym_members_exercise_tracking.csv'

if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    print(f'✅ Dataset loaded: {len(df)} records')
    print(df.head())
else:
    print('❌ CSV file not found. Please upload gym_members_exercise_tracking.csv')
```

```
def gym_ai_assistant(user_query):
    """Gym AI Assistant that answers questions based on member data."""
    csv_path = '/content/gym_members_exercise_tracking.csv'
    
    if not os.path.exists(csv_path):
        return 'Error: Dataset not found. Please upload gym_members_exercise_tracking.csv'
    
    df = pd.read_csv(csv_path)
    
    # Create data context
    context = f'Gym Dataset: {len(df)} members. '
    context += f'Avg Calories: {df["Calories_Burned"].mean():.2f}. '
    context += f'Avg Session: {df["Session_Duration (hours)"].mean():.2f} hours.'
    
    prompt = f'Context: {context}\nQuestion: {user_query}\nAnswer as a professional gym assistant.'
    
    try:
        response = ai.generate_text(prompt)
        return response
    except Exception as e:
        return f'Error: {e}'

# Test the assistant
print(gym_ai_assistant('What is the average calories burned per session?'))
```
